# Exploratory Star-List Coordinate Matching

This notebook is a small historical experiment using sample lists distributed with the third-party MATCH astronomy software. It plots two lists and applies a recorded affine transformation to one list for visual comparison.

**Scope:** this is an exploratory visualization, not a complete star-identification algorithm or spacecraft-attitude solution. The sample data are MATCH test fixtures, not measurements collected by the author.

```python
import matplotlib.pyplot as plt
from astropy.io import ascii

%matplotlib inline
```

```python
fig, ax = plt.subplots(figsize=(10, 8))
ax.set_xlabel('X coordinate (sample data units)')
ax.set_ylabel('Y coordinate (sample data units)')
ax.set_title('Sample star lists and an affine coordinate transformation')

data1 = ascii.read('../third_party/MATCH/testdata/selfa.dat')
x1 = data1['col2']
y1 = data1['col3']
ax.plot(x1, y1, 'rs', label='MATCH sample list A')

data2 = ascii.read('../third_party/MATCH/testdata/selfb.dat')
x2 = data2['col2']
y2 = data2['col3']
ax.plot(x2, y2, 'g+', markersize=10, label='MATCH sample list B')

# Recorded affine-transformation coefficients from the original experiment.
a = 4.366142173e-02
b = 7.074846726e-01
c = -7.073670562e-01
d = 7.763385459e-02
e = 7.080665130e-01
f = 7.068801461e-01

x_transformed = a + b * x1 + c * y1
y_transformed = d + e * x1 + f * y1
ax.plot(x_transformed, y_transformed, 'kd', markersize=6, label='List A after recorded transform')

ax.legend()
ax.grid(True, alpha=0.25)
fig.tight_layout()
plt.show()
```
