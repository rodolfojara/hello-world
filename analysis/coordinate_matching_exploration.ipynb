# Coordinate-matching exploration

This notebook visualizes the **synthetic catalogs included in MATCH's own self-test data**. It plots catalog A, catalog B, and catalog A after applying a sample coordinate transformation. These are test fixtures, not real satellite observations.

The transformation coefficients below are example values recorded from an exploratory MATCH run; this notebook does not run MATCH or fit the coefficients.

```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt

# Support execution with the working directory set to either the repository root or analysis/.
if Path("third_party/match/selfa.dat").exists():
    DATA_DIR = Path("third_party/match")
else:
    DATA_DIR = Path("../third_party/match")

catalog_a = np.loadtxt(DATA_DIR / "selfa.dat")
catalog_b = np.loadtxt(DATA_DIR / "selfb.dat")

# The files contain object ID, x coordinate, y coordinate, and magnitude.
x_a, y_a = catalog_a[:, 1], catalog_a[:, 2]
x_b, y_b = catalog_b[:, 1], catalog_b[:, 2]
```

```python
# Example coefficients recorded from an exploratory MATCH run.
a = 4.366142173e-02
b = 7.074846726e-01
c = -7.073670562e-01
d = 7.763385459e-02
e = 7.080665130e-01
f = 7.068801461e-01

x_a_transformed = a + b * x_a + c * y_a
y_a_transformed = d + e * x_a + f * y_a

fig, ax = plt.subplots(figsize=(8, 7))
ax.scatter(x_a, y_a, marker="s", label="Catalog A (synthetic)")
ax.scatter(x_b, y_b, marker="+", s=90, label="Catalog B (synthetic)")
ax.scatter(x_a_transformed, y_a_transformed, marker="D", label="Catalog A transformed")
ax.set(title="MATCH self-test catalogs: example coordinate transformation",
       xlabel="X coordinate", ylabel="Y coordinate")
ax.set_aspect("equal", adjustable="datalim")
ax.grid(True, alpha=0.25)
ax.legend()
plt.show()
```
